## 4.5 Principal Component Analysis

Principal component analysis (PCA) is a classical method for reducing the number of attributes in the data by projecting the data from its original high-dimensional space into a lower-dimensional space. The new attributes (also known as components) created by PCA have the following properties: (1) they are linear combinations of the original attributes, (2) they are orthogonal (perpendicular) to each other, and (3) they capture the maximum amount of variation in the data.

Apply PCA to an image dataset containing 16 RGB files of 111 x 111 pixels. First, read and display every image, flatten each one into 111 x 111 x 3 = 36,963 feature values, and store the results in a 16 x 36,963 data matrix.

The image data matrix $X$ has 16 rows (images) and 36,963 columns (pixel-channel features). PCA transforms this matrix through the following conceptual steps:

1. **Center the data.** Subtract the mean of each feature from that feature's values. This produces the centered matrix $X_c$.
2. **Define the covariance matrix.** Compute $C = X_c^T X_c / (n-1)$. Here, $C$ has shape 36,963 x 36,963, and each entry describes how two pixel-channel features vary together across the images.
3. **Find the eigenvectors.** Compute the eigenvectors and eigenvalues of $C$. Each eigenvector defines a direction in the original 36,963-dimensional feature space. Its eigenvalue measures how much data variance lies along that direction.
4. **Define the new coordinate system.** Order the eigenvectors by decreasing eigenvalue. The first eigenvector defines the first principal-component direction (PC1), the second defines PC2, and so on. These mutually orthogonal directions form the axes of the PCA space.
5. **Project the images.** Let $v_1$ and $v_2$ be the first two eigenvectors, each containing 36,963 coefficients. For a centered, flattened image $x_i$, compute two dot products: $z_{i1} = x_i · v_1$ and $z_{i2} = x_i · v_2$. These scalar values measure how far the image lies along the PC1 and PC2 directions. The image is therefore mapped from 36,963 values to the two-dimensional point $(z_{i1}, z_{i2})$. Performing this operation for all 16 images can be written as $Z = X_c W$, where $W$ contains $v_1$ and $v_2$ as its columns. The matrix shapes are $(16 × 36{,}963)(36{,}963 × 2) = (16 × 2)$.

The eigenvectors themselves contain 36,963 coefficients, whereas the `pc1` and `pc2` columns contain the 16 images' projected coordinates along those directions. Scikit-learn normally performs this computation efficiently using singular value decomposition (SVD), without explicitly constructing the very large covariance matrix.
